# Zepto Delivery Performance Analytics & Machine Learning Pipeline
This Jupyter notebook provides an end-to-end exploratory data analysis (EDA), data cleaning, feature engineering, model training, comparison, evaluation, and inference workflow for **Zepto Quick-Commerce Delivery Analytics**.

```python
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from analytics.preprocessing import DataPreprocessor
from analytics.visualize import AnalyticsVisualizer
from analytics.train import ModelTrainer
from analytics.evaluate import ModelEvaluator
from analytics.predict import DeliveryPredictor

print('Data science libraries imported successfully!')
```

## 1. Data Loading & Initial Inspection

```python
processor = DataPreprocessor()
df_raw = processor.load_data()
print('Raw Dataset Shape:', df_raw.shape)
df_raw.head()
```

## 2. Data Preprocessing & Outlier Handling

```python
df_clean = processor.clean_and_impute(df_raw)
df_capped = processor.handle_outliers(df_clean)
df_capped.describe()
```

## 3. Exploratory Data Analysis & Visualizations

```python
visualizer = AnalyticsVisualizer()
visualizer.plot_histogram(df_capped)
visualizer.plot_heatmap(df_capped)
plt.show()
```

## 4. Model Training & Comparison

```python
trainer = ModelTrainer()
leaderboard, best_model_name, artifact = trainer.train_and_evaluate_all()
pd.DataFrame(leaderboard).T
```

## 5. Model Evaluation & ROC / Confusion Matrix

```python
evaluator = ModelEvaluator()
metrics = evaluator.evaluate_model()
print('Best Model Accuracy:', metrics['accuracy'])
print('Best Model F1 Score:', metrics['f1_score'])
```

## 6. Real-Time Inference Demo

```python
predictor = DeliveryPredictor()
sample = {
    'customer_tenure_months': 6,
    'order_distance_km': 7.2,
    'item_count': 12,
    'order_value_inr': 1850.0,
    'traffic_density': 'High',
    'weather_condition': 'Rainy',
    'driver_experience_years': 1,
    'delivery_time_mins': 24.5
}
result = predictor.predict_sample(sample)
print('Prediction Output:', result)
```
